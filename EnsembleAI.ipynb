# EnsembleAI

Find out who in your ensemble is out of tune, and when.

**Setup:** Runtime → Change runtime type → **T4 GPU**. Add two secrets (key icon on the left): `HF_TOKEN` (Hugging Face read token with access to `facebook/sam-audio-small`) and `MODEL_API_KEY` (Meta Model API key). Upload `ensembleai_core.py`, `ensembleai_app.py`, `cq.wav`, and `cq_detuned.wav` to the Files panel.

### 1. Install
The session restarts itself when this finishes. That's expected; your files stay.

```
!pip install -q git+https://github.com/facebookresearch/sam-audio.git openai
import os; os.kill(os.getpid(), 9)   # auto-restart so the new packages load
```
